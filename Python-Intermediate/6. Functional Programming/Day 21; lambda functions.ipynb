```python
# Simple Lambda function
add= lambda x,y : x+y
print(add(2,3))
```

```python
# def v lambda
def mult_def(a,b):
  return a*b
print(mult_def(2,5))


mult_lam = lambda a,b : a*b
print(mult_lam(2,5))
```

```python
# Lambda with filter function
print(filter.__doc__, '\n')

a= [ x for x in range(1,10,1)]

b= [ x for x in range(11,20)]

even_a = filter(lambda x: x%2==0, a)
print(list(even_a))

even_b = filter(lambda x: x%2==0, b)
print(list(even_b))
```

```python
# Lambda with map function
print(map.__doc__, '\n')

radius = [1,2,3, 4, 10]
area = map(lambda x: 3.14*x*x , radius)
print(list(area))
```

```python
# Lambda with reduce function
from functools import reduce
print(reduce.__doc__, '\n')

a= [x for x in range(1,5)]
factorial = reduce(lambda x,y: x*y, a)
print(factorial)
```

```python
# 🧩 Problem: Filter and Transform Numbers

numbers = [1, 5, 8, 12, 15, 20, 23, 30, 7, 10]

even_numbers = filter(lambda x:x%2==0, numbers)
print('Even numbers :::',list(even_numbers))

greater_than_10 = filter(lambda x: x>10, numbers)
print('Numbers greater than 10 :::',list(greater_than_10))

square = map(lambda x: x**2, numbers)
print('Square :::', list(square))

# Finding squares of even numbers only

even_numbers_b = list(filter(lambda x:x%2==0, numbers))
square_b = list(map(lambda x: x**2, even_numbers_b))
print('Square of even numbers :::', square_b)
```

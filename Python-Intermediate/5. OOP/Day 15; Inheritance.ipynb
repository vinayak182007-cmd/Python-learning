```python
# Basic

class Animal:
  def __init__(self, name):
    self.name= name

  def info(self):
    print(f'animal bread is {self.name}')

class Dog(Animal):
  def sound(self):
    print(self.name, "barks!!")

d= Dog("Buddy")

d.info()
d.sound()
```

```python
# Using super() and MRO

# Parent class
class Animal:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def category(self):
    if self.age>5:
      print("Old")
    else:
      print("Young")

# Child class
class Lion(Animal):
  def __init__(self, name , breed, age):
    super().__init__(name, age)     # Using name and age from parent class based on MRO
    self.breed= breed

  def sound(self):
    print(f'{self.name} is a Lion and it ROARS')

T2= Lion('T2', 'Asian', 4)
T2.category()
T2.sound()
```

```python
# Method Overriding in Inheritence

class startup:
  def category(self):
    print('Technology')

class edtech(startup):
  def category(self):
    print('Combining technology with education')

unacademy = edtech()
unacademy.category()
```

Types of Inheritance

```python
# 1. Single Inheritence

class A:
  def fun1(self):
    print("Parent class")

class B(A):
  def fun2(self):
    print("Child Class")

obj= B()
obj.fun1()
obj.fun2()
```

```python
# 2. Multiple Inheritence

class Mother:
  def trait_from_mother(self):
    a= ['fair color','sharpe nose','curly hairs']
    print(f'Traits from Mother : {a}')

class Father:
  def trait_from_father(self):
    b= ['Tall hight','Muscular body']
    print(f'Traits from Father : {b}')

class Child(Mother, Father):
  def trait(self):
    print('Child contains traits of mother and father')

Rohan = Child()
Rohan.trait_from_father()
Rohan.trait_from_mother()
```

```python
# 3. Multilevel Inheritence

class Grandfather:
  def __init__(self, gf_name):
    self.gf_name = gf_name

class Father(Grandfather):
  def __init__(self, f_name, gf_name):
    self.f_name = f_name
    super().__init__(gf_name)

class Son(Father):
  def __init__(self, s_name, f_name, gf_name):
    self.s_name = s_name
    super().__init__(f_name, gf_name)

  def print_name(self):
    print(f'Name of Grandfather of {self.s_name} is {self.gf_name}')
    print(f'Name of Father of {self.s_name} is {self.f_name}')

s1 = Son('Rohan','Mohan','Pavan')
s1.print_name()
```

```python
# 4. Hierarchical Inheritance

class Parent:
    def func1(self):
        print("This function is in parent class.")

class Child1(Parent):
    def func2(self):
        print("This function is in child 1.")

class Child2(Parent):
    def func3(self):
        print("This function is in child 2.")

object1 = Child1()
object2 = Child2()
object1.func1()
object1.func2()
object2.func1()
object2.func3()
```

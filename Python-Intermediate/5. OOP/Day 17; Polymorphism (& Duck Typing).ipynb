```python
# Runtime Polymorphism (method overriding)

class payment:
  def payment_process(self):
    return "Processing general payement..."

class CreditCard(payment):
  def payment_process(self):
    return "Processing payment from Credit Card"

class UPIPayment(payment):
  def payment_process(self):
    return "Processing payment from UPI method"

# Defining make_payment function outside any class

def make_payment(payment_obj, amount):
  method= payment_obj.payment_process()
  return f"Detail: {method} and amount = {amount} rs"

pay= CreditCard()
print(make_payment(pay, 500))
```

```python
# Above example with : Runtime polymorphism + inheritence

class payment:
  def payment_process(self):
    return "Processing general payement..."

  def make_payment(self, amount):
    msg= self.payment_process()
    return f"amount {amount} rs : {msg}"

class CreditCard(payment):
  def payment_process(self):
    return "Processing payment from Credit Card..."

class UPIPayment(payment):
  def payment_process(self):
    return "Processing payment from UPI method"

pay = CreditCard()
pay.make_payment(500)
```

```python
# Duck Typing

class PDFfile:
  def print_content(self):
    return "Printing PDF document ..."

class WordFile:
  def print_content(self):
    return "Printing Word File ..."

class imgfile:
  def print_content(self):
    return "Printing Image File ..."

def start_printing(file_obj):
  print(file_obj.print_content())

print_1 = WordFile()
start_printing(print_1)
```

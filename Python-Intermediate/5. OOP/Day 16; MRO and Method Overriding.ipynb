```python
# Method Resolution Order in Diamond Problem

class C1:
  def m(self):
    print('Message from C1')

class C2(C1):
  def m(self):
    print('Message from C2')
    super().m()

class C3(C1):
  def m(self):
    print('Message from C3')
    super().m()

class C4(C2, C3):
  def m(self):
    print('Message from C4')
    super().m()

obj= C4()
obj.m()

print(C4.mro())
```

```python
# Calling methods of parent classes from child class

class Class1:
    def m(self):
        print("In Class1")

class Class2(Class1):
    def m(self):
        print("In Class2")
        Class1.m(self)

class Class3(Class1):
    def m(self):
        print("In Class3")
        Class1.m(self)

class Class4(Class2, Class3):
    def m(self):
        print("In Class4")
        Class3.m(self)
        Class2.m(self)

obj = Class4()
obj.m()
```

```python
# Method Overriding with Constructors

class Emp:
  def __init__(self):
    self.role= "Employee"

  def display(self):
    print('Role= ', self.role)

class Manager(Emp):
  def __init__(self):
    super().__init__()
    self.role= "Manager"

  def display(self):
    print('Role= ', self.role)

e1 = Emp()
e2 = Manager()
e1.display()
e2.display()
```

```python
class Report:
    def generate(self):
        print("Generating report...")
        self.display()

    def display(self):
        print("Displaying basic report")

class SalesReport(Report):
    def display(self):
        print("Displaying sales report")

obj1 = SalesReport()
obj1.generate()

obj2 = Report()
obj2.generate()
```

```python
# Calling Parent Method Using Class Name

class Employee:
  def calc_salary(self, basic_pay= 30000):
    return basic_pay

class Manager(Employee):
  def calc_salary(self):
    basic= Employee.calc_salary(self)
    allowance= 10000
    print('Manager salary=',basic + allowance)

Manager1 = Manager()
Manager1.calc_salary()
```

```python
class Employee:
  def calc_salary(self):
    self.basic_pay= 30000

class Manager(Employee):
  def calc_salary(self):
    Employee.calc_salary(self)
    allowance= 10000
    print('Manager salary=',self.basic_pay + allowance)

Manager1 = Manager()
Manager1.calc_salary()
```

```python
# Calling Parent Method Using super()

class Order:
  def calculate_charge(self):
    self.basic_charge = 50

class PremiunOrder(Order):
  def calculate_charge(self):
     super().calculate_charge()
     print("Priority delivery Charges= ", self.basic_charge + 20, "Rupees")

Order1 = PremiunOrder()
Order1.calculate_charge()
```

```python
# Inheritance

class Animal:
  def sound(self):
    print('Animals make sound')

class Dog(Animal):
  def legs(self):
    print('Dog has 4 legs')

dog= Dog()
dog.sound()
dog.legs()
```

```python
# inheritence with __init__() method

class A:
  def __init__(self):
    print('A is called')

class B(A):
  def __init__(self):
    super().__init__()     # Call parent __init__
    print("B is called")

obj = B()
```

```python
# Use of __str__ method
# Without __str__

class Cat:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        print(f'Name of cat is {self.name} having age {self.name}')

cat1 = Cat("Buddy", 3)
cat2 = Cat("Charlie", 5)

print(cat1)
print(cat2)
```

```python
# With __str__

class Kitty:
  def __init__(self, name, age):
    self.name= name
    self.age = age

  def __str__(self):
    return f'Name of cat is {self.name} having age {self.name}'

kitty1 = Kitty('Buddy', 3)

print(kitty1)
```

```python
# Class vs Instance varible

class inflation:
  measured_using= "WPI"       # Class variable

  def __init__(self, country, rate):
    self.country= country     # Instance variable
    self.rate= rate           # Instance variable

Inflation_1 = inflation("India", 4)
Inflation_2 = inflation("Japan", 5)
Inflation_3 = inflation("USA", 5)

print(Inflation_1.measured_using)
print(Inflation_2.measured_using)
print(Inflation_3.measured_using)
```

```python
# Updating Class variable

# method 1: Modifying via instance variable
Inflation_2.measured_using= "CPI"

print(Inflation_1.measured_using)
print(Inflation_2.measured_using)
print(Inflation_3.measured_using)
```

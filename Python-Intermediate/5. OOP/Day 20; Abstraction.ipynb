```python
from abc import ABC, abstractmethod

class Animal(ABC):
  @abstractmethod
  def species(self):
    pass                   # Abstract method, no implementation

  @abstractmethod
  def legs(self):
    pass                  # Abstract method, no implementation

  def make_sound(self):
    return "General Sound"       # Concrete method, have implementation


class Dog(Animal):
  def species(self):      # Must to have (due to @abstractmethod in parent class)
    return "Bull Dog"
  def legs(self):         # Must to have (due to @abstractmethod in parent class)
    return 4
  def name(self):         # Additional methods are permitted
    return "Tiger"

tiger = Dog()
print(tiger.species())
print(tiger.name())
print(tiger.make_sound())

# animal2 = animal() -----> error, abstract class method lacks implementation (& incomplete)
```

```python
# Use of @property
# Here, we are using above abstract class, Animal

class Cat(Animal):
  @property
  def species(self):
    return "Kitten"
  def legs(self):
    return 4

kitty = Cat()
#print(kitty.species())
print(kitty.species) # Due to `@property` method `species` behaves like attribute
# kitty.species() -----> error
```

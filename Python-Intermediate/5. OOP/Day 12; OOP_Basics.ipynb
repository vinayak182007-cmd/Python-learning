```python
# Basics

class Car:
  def __init__(self):
    self.color = ''
    self.model = ''

  def start(self):
    print(f"{self.model} is started")

Car.model = 'A12'
print(Car.model)

myCar= Car()
myCar.model = "Toyota"
myCar.start()

#Create second car object
mySecondCar = Car()
mySecondCar.color= "Black"
mySecondCar.model = "Mahindra"

# __init__ = sets up a new object when you create it
# self = the particular object you're currently working with
```

```python
print("My car color:", myCar.color)
mySecondCar.start()
```

```python
class data_store:

    def __init__(self):
        self.a = 'a Inside'           # Public
        self._b = '_b Inside'         # Protected (convention)
        self.__c = '__c Inside'       # Private

hello = data_store()

hello.a = 'a Outside'
hello._b = '_b Outside'
hello.__c = '__c outside'     # This does NOT modify the original __c

print(hello.a)
print(hello._b)
print(hello.__c)
```

```python
# Data Encapsulation

class data_store2:
  def __init__(self):
     self.a= 'Hello a!'
     self._b= 'Hello _b!'
     self.__c= 'Hello __c!'

obj2 = data_store2()
print(obj2.a)
print(obj2._b)
print(obj2._data_store2__c)
# print(obj2.__c) = no attribute error due to NAME MANDLING
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
    @abstractmethod
    def legs(self):
	    pass

class Dog(Animal):
    def sound(self):
        print("Dog barks")
    def legs(self):
		    print('Dog has 4 legs')

class Cat(Animal):
    def sound(self):
        print("Cat meows")
    def legs(self):
        print("Cat has 4 legs")

d = Dog()
c = Cat()

d.sound()
c.sound()
```

```python
class A:
  def __init__(self, num_1):
    self.num_1 = num_1

  def __sub__(self, other):
    return self.num_1 - other.num_1

obj1 = A(5)
obj2 = A(4)

print(obj1 - obj2)
print(A.__sub__(obj1, obj2))
print(obj1.__sub__(obj2))
```

```python
# Time Duration Calculator using Operator Overloading and Magic Method

class Duration:
  def __init__(self, d1):
    self.d1 = d1

  def __add__(self, other):
    return f"{(self.d1 + other.d1)//60} hours, {(self.d1 + other.d1)% 60}"

dur1= Duration(50)
dur2= Duration(20)
print(dur1 + dur2)
```

```python
class B:
  def __init__(self, num_1):
    self.num_1 = num_1

  def __add__(self, other, other_2):
    return self.num_1 + other.num_1 + other_2.num_1

obj1 = B(1)
obj2 = B(2)
obj3 = B(3)

print(obj1.__add__(obj2, obj3))
```

```python
# Calculating Complex number

class Complex:
  def __init__(self, real, img):
    self.real = real
    self.img = img

  def __add__(self, other):
    return self.real + other.real , self.img + other.img

complex_1 = Complex(2,5)
complex_2 = Complex(3,4)

print(complex_1 + complex_2)
```

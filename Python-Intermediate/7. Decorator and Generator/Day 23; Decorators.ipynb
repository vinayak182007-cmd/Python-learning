```python
# Making hands dirty of decorators

def my_deco(func):
  def wrapper(*args, **kwargs):
    print('Initialization....')
    func(*args, **kwargs)
    print('Termination....')
  return wrapper

@my_deco
def counting_till(n):
  for i in range(1,n):
    print(i)

counting_till(5)
```

### Types of Decorators

```python
# 1. Functional Decorators
# Above code 'counting_till' is example of this

# 2. Method decorators

def log_method(func):
  def wrapper(self, *arg, **kwargs):
    print(f'Calling {func.__name__}...')
    print(f'Finished {func.__name__}...')
    store_result_from_class_method = func(self, *arg, **kwargs)
    return store_result_from_class_method
  return wrapper

class Calculator:
    @log_method
    def add(self, a, b):
        return a + b
    @log_method
    def multiply(self, a, b):
        return a * b

calc = Calculator()
print(calc.add(3, 5))
print(calc.multiply(4, 2))
```

### Built-in Decorators

```python
# 1. @staticmethod : define a method that doesn't operate on an instance of class

class maths:
  @staticmethod
  def add_mult(a,b,c):
    return (a+b)*c

obj1 = maths()
obj1.add_mult(1,2,3)
```

```python
# 2. @classmethod : @classmethod Python ko bolta hai: "Is method ko object nahi, class ke context me run karo."

class Employee:
    raise_amount = 1.05
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    @classmethod
    def set_raise_amount(cls, amount):
        cls.raise_amount = amount

Employee.set_raise_amount(1.10)
print(Employee.raise_amount)
```

```python
# 3. @property : When it is required to use method as attribute

class circle:
  def __init__(self, radius):
    self.radius = radius

  @property
  def area(self):
    return 3.14*(self.radius**2)

c1 = circle(7)
print(c1.area)        # here, we are using c1.area instead of c1.area()
```

```python
# Basics

def my_generator():
  print('First number...')
  yield 12
  print('second number ...')
  yield 34

gen= my_generator()
print(gen)        # generator object
print(next(gen))  # First number
print(next(gen))  # Second number
#print(next(gen)) Error

print('---'*30)

for value in my_generator():    # Using for loop
  print(value)                  # Atomatically handle <StopIteration> error
```

```python
def count_up():
    print("Start")
    yield 1
    print("Middle")
    yield 2
    print("End")

counter = count_up()

print("--- Call 1 ---")
a = next(counter)   # print code before `yeild 1` then assign a=1
# print(a)          # Output= 1 because now a=1

print("--- Call 2 ---")
b = next(counter)   # print code before `yeild 2` then assign B=2
# print(b)          # Output= 2 because now b=2
```

### Advance Generator methods

1. send()

```python
# 1. send()

def my_receiver():
    print("Generator start")
    while True:             # <<<imp>>> this loop is used to prevent unusual StopIteration error
      received_val = yield
      print(f"Message Recived: {received_val}")

gen = my_receiver()
next(gen)         # Priming = start generator and pause at yield

gen.send("Hello Python!")
```

```python
def multiple_reciver():
  val1 = yield "Ready 1"
  print(f"Received: {val1}")
  val2 = yield "Ready 2"
  print(f"Received: {val2}")

gen= multiple_reciver()
next(gen)
print(gen.send("Hello"))
# next(gen)
# print(gen.send("Python"))  ----> StopIteration error: Because next(gen) ke baad generator ka code finish ho jata hai
```

2. close()

```python
def value_stream():
  for i in range(100):
    yield i

gen= value_stream()
print(next(gen))
print(next(gen))
print(next(gen))
gen.close()
# print(next(gen)) ----> StopIteration error
```

```python
# With try expect

def my_stream():
  try:
    yield 1
    yield 2
  except GeneratorExit:
    print("Generator closed")

gen= my_stream()
print(next(gen))
gen.close()
```

3. yield from

```python
from re import I
def numbers():
  for i in range(3):
    yield i

def letters():
  for i in "Python":
    yield i

def combination():
  yield from numbers()
  yield from letters()

for item in combination():
  print(item)
```

### Practice problem

Temperature Monitoring

```python
def temp_monitor():
  try:
    temp = yield
    while True:
      if temp>40:
        temp= yield 'High temperature alert'
      else:
        temp= yield 'Temperature in control'
  except GeneratorExit:
    print('Sensor shutting down')

sensor = temp_monitor()

next(sensor)     # Priming

print(sensor.send(40))
print(sensor.send(44))

sensor.close()
```
